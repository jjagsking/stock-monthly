```python
import yfinance as yf
import pandas as pd
import calendar
import html
import time
from IPython.display import display, HTML

# 👇 조회할 연월만 수정하세요
YEAR = 2013
MONTH = 8

if not (1 <= MONTH <= 12 and 1990 <= YEAR <= 2100):
    raise ValueError("YEAR/MONTH 값을 확인하세요.")

current = pd.Timestamp(YEAR, MONTH, 1)
previous = current - pd.DateOffset(months=1)
next_month = current + pd.DateOffset(months=1)
all_assets = BASE + EXTRA
tickers = [t for t, _ in all_assets]

def get_monthly_close(frame, ticker):
    """Yahoo 월봉 Close에서 직전월/당월 종가를 찾습니다."""
    if frame is None or frame.empty:
        return None, "데이터 없음"
    try:
        if isinstance(frame.columns, pd.MultiIndex):
            if ticker in frame.columns.get_level_values(0):
                closes = frame[ticker]["Close"]
            elif ticker in frame.columns.get_level_values(1):
                closes = frame["Close"][ticker]
            else:
                return None, "데이터 없음"
        else:
            closes = frame["Close"]
        closes = pd.to_numeric(closes, errors="coerce").dropna()
        closes.index = pd.to_datetime(closes.index).tz_localize(None).to_period("M")
        closes = closes.groupby(level=0).last()
        p, c = previous.to_period("M"), current.to_period("M")
        if p not in closes.index or c not in closes.index:
            return None, "상장 전 또는 데이터 없음"
        if closes.loc[p] <= 0:
            return None, "데이터 오류"
        return (float(closes.loc[c]) / float(closes.loc[p]) - 1) * 100, ""
    except Exception as exc:
        return None, f"조회 오류: {str(exc)[:60]}"

# 한 번에 너무 많은 종목을 요청하지 않도록 묶어서 조회합니다.
values = {}
for i in range(0, len(tickers), 25):
    batch = tickers[i:i+25]
    print(f"조회 중: {i+1}~{min(i+25,len(tickers))}/{len(tickers)}", end="\r")
    try:
        data = yf.download(
            batch, start=previous.strftime("%Y-%m-%d"),
            end=next_month.strftime("%Y-%m-%d"),
            interval="1mo", auto_adjust=False,
            group_by="ticker", progress=False,
            threads=True, timeout=25
        )
    except Exception:
        data = None
    for ticker in batch:
        values[ticker] = get_monthly_close(data, ticker)

# 일괄 조회 실패 종목은 개별 재시도
for ticker in tickers:
    if values[ticker][0] is not None:
        continue
    try:
        data = yf.download(
            ticker, start=previous.strftime("%Y-%m-%d"),
            end=next_month.strftime("%Y-%m-%d"),
            interval="1mo", auto_adjust=False,
            progress=False, threads=False, timeout=15
        )
        retry = get_monthly_close(data, ticker)
        if retry[0] is not None:
            values[ticker] = retry
    except Exception:
        pass

# 코스피200: Yahoo ^KS200가 누락될 때 이전 노트북과 동일하게 FDR KS200 조회
ks200_source = "Yahoo Finance"
if values.get("^KS200", (None, ""))[0] is None:
    try:
        import FinanceDataReader as fdr
        ks = fdr.DataReader("KS200", previous.strftime("%Y-%m-%d"), next_month.strftime("%Y-%m-%d"))
        if ks is not None and not ks.empty:
            ks = ks.copy()
            ks.index = pd.to_datetime(ks.index).to_period("M")
            monthly = ks.groupby(level=0)["Close"].last()
            p, c = previous.to_period("M"), current.to_period("M")
            if p in monthly.index and c in monthly.index and monthly.loc[p] > 0:
                values["^KS200"] = ((float(monthly.loc[c]) / float(monthly.loc[p]) - 1) * 100, "")
                ks200_source = "FinanceDataReader KS200 (Yahoo 데이터 누락으로 대체)"
            else:
                ks200_source = "조회 실패 (Yahoo/FDR 모두 데이터 없음)"
        else:
            ks200_source = "조회 실패 (Yahoo/FDR 모두 데이터 없음)"
    except Exception as e:
        ks200_source = "조회 실패: " + str(e)[:80]

rows = []
for category, assets in [("기본", BASE), ("추가", EXTRA)]:
    for ticker, name in assets:
        pct, status = values[ticker]
        rows.append({"분류": category, "티커": ticker, "종목명": name, "등락률값": pct, "상태": status})

df = pd.DataFrame(rows)
base_df = df[df["분류"] == "기본"]
extra_df = df[df["분류"] == "추가"].sort_values("등락률값", ascending=False, na_position="last", kind="stable")
result = pd.concat([base_df, extra_df], ignore_index=True)
result["월간 등락률"] = result["등락률값"].map(lambda x: f"{x:+.1f}%" if pd.notna(x) else "—")

# Colab 휴대폰 화면에 맞춘 3열 테이블: 구분/상태 열은 숨김
tr = []
for _, r in result.iterrows():
    pct = r["등락률값"]
    bg, fg = ("#fff0f0", "#c62828") if pd.notna(pct) and pct > 0 else (("#edf4ff", "#1565c0") if pd.notna(pct) and pct < 0 else ("#f4f4f4", "#666666"))
    status = html.escape(str(r["상태"] or ""), quote=True)
    tr.append(f'<tr style="background:{bg};color:{fg}" title="{status}"><td>{html.escape(r["티커"])}</td><td>{html.escape(r["종목명"])}</td><td class="pct">{r["월간 등락률"]}</td></tr>')

markup = f"""
<div style="max-width:100%;overflow-x:hidden;font-family:Arial,'Noto Sans KR',sans-serif">
<div style="font-size:14px;margin:8px 0;font-weight:bold">{YEAR}년 {MONTH}월 · 전체 {len(result)}개 · 확인 {int(df['등락률값'].notna().sum())}개</div>
<table style="width:100%;border-collapse:collapse;table-layout:fixed;font-size:13px;line-height:1.3">
<thead><tr style="background:#eee;color:#222"><th style="width:26%">티커</th><th style="width:46%">종목명</th><th style="width:28%;text-align:right">등락률</th></tr></thead>
<tbody>{''.join(tr)}</tbody></table></div>
<style>table td,table th{{padding:8px 4px;overflow-wrap:anywhere;text-align:left;border-bottom:1px solid #ddd}}table td.pct{{text-align:right;white-space:nowrap;font-weight:600}}</style>
"""
display(HTML(markup))
print("코스피200 출처:", ks200_source)
print("조회 불가 종목:", ", ".join(result.loc[result["등락률값"].isna(), "티커"].tolist()) or "없음")
# 필요하면 CSV 저장:
# result[["티커", "종목명", "월간 등락률"]].to_csv(f"monthly_{YEAR}_{MONTH:02d}.csv", index=False, encoding="utf-8-sig")
```
